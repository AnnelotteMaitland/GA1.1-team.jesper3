So, my function is subtract(a,b) it will take two values, such as a and b, and as the output it will present the result in the form of "When you subtract b from a you get: ..."

```python
def subtract(a,b):
    c = a - b
    return c
```

```python
def distance_from_zero(x):
    r = abs(subtract(x,0))
    return print(f'the distance from zero to {x} is {r}')


```

```python
distance_from_zero(-5)
```
